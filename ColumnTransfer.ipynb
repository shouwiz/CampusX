```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df = pd.read_csv('covid_toy.csv')
df
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(df.drop(columns=['has_covid']),df['has_covid'],test_size=0.2)
```

```python
df.isnull().sum()
```

```python
X_train
```

#### Normal Scenario

```python
si = SimpleImputer()

X_train_fever = si.fit_transform(X_train[['fever']])
X_test_fever = si.fit_transform(X_test[['fever']])
X_train_fever.shape
```

```python
oe = OrdinalEncoder(categories=[['Mild','Strong']])
X_train_cough = oe.fit_transform(X_train[['cough']])
X_test_cough = oe.fit_transform(X_test[['cough']])
X_train_cough.shape
```

```python
ohe = OneHotEncoder(drop='first',sparse_output=False)
X_train_gender_city = ohe.fit_transform(X_train[['gender','city']])
X_test_gender_city = ohe.fit_transform(X_test[['gender','city']])
X_train_gender_city.shape
```

```python
X_train_age = X_train.drop(columns=['gender','city','fever','cough',])
X_test_age = X_test.drop(columns=['gender','city','fever','cough',])
X_train_age.shape
```

#### Column Transformer

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer = ColumnTransformer(transformers=[('tf1',SimpleImputer(),['fever']),
                                              ('tf2',OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),
                                              ('tf3',OneHotEncoder(drop='first',sparse_output=False),['gender','city'])],
                                              remainder = 'passthrough')
```

```python
transformer.fit_transform(X_train).shape
```

```python
transformer.transform(X_test).shape
```
